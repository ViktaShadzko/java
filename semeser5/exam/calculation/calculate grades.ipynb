```python
import pandas as pd



def calculate_final_grade(exam_grade, lab1, lab2, lab3, task1, task2, task3, lab4):
    """
    Calculate final grade using the formula:
    0.3*(Exam Grade) + 0.7*0.66*((Lab1 + Lab2 + Lab3)/3) + 0.7*0.34*((Task1 + Task2 + Task3)/3) + 0.05*(Lab4)

    Parameters:
    - exam_grade: Оценка за экзамен (0-100)
    - lab1, lab2, lab3: Оценки за лабораторные работы (0-100)
    - task1, task2, task3: Оценки за задания (0-100)
    - lab4: Балл за выполнение Lab 4 (0-100)

    Returns:
    - final_grade: Итоговая оценка (0-100)
    """
    # Calculate average of Labs 1-3
    labs_avg = (lab1 + lab2 + lab3) / 3

    # Calculate average of Tasks 1-3
    tasks_avg = (task1 + task2 + task3) / 3

    # Apply the formula
    final_grade = (
            0.3 * exam_grade +
            0.7 * 0.66 * labs_avg +
            0.7 * 0.34 * tasks_avg +
            0.05 * lab4
    )

    return final_grade


# Example usage for a single student
exam = 85
lab1 = 90
lab2 = 85
lab3 = 88
task1 = 75
task2 = 80
task3 = 78
lab4 = 95

final = calculate_final_grade(exam, lab1, lab2, lab3, task1, task2, task3, lab4)
print(f"Example Student Final Grade: {final:.2f}")
```

```python
# Calculate grades for multiple students using pandas
# Create sample data (replace with actual student data)

students_data = {
    'Student': ['Banshik', 'Alina', 'Maksim', 'Yaroslav', 'Olya', 'Daniil', 'Kirill'],
    'Exam':  [7, 9, 6 , 6, 9 , 6, 6],
    'Lab1':  [8, 9, 8 , 5, 9 , 8, 6],
    'Lab2':  [7, 7, 8 , 5, 9 , 7, 5],
    'Lab3':  [8, 8, 8 , 5, 9 , 7, 5],
    'Task1': [8, 8, 10, 8, 10, 8, 7],
    'Task2': [9, 8, 9 , 6, 10, 8, 8],
    'Task3': [8, 7, 9 , 5, 9 , 7, 5],
    'Lab4':  [0, 0, 0 , 0, 0 , 0, 0]
}

df = pd.DataFrame(students_data)

# Calculate final grades for all students
df['Final_Grade'] = df.apply(
    lambda row: calculate_final_grade(
        row['Exam'], row['Lab1'], row['Lab2'], row['Lab3'],
        row['Task1'], row['Task2'], row['Task3'], row['Lab4']
    ),
    axis=1
)

# Round to 2 decimal places
df['Final_Grade'] = df['Final_Grade'].round(2)

print("\nGrades Table:")
print(df.to_string(index=False))

# Calculate statistics
print(f"\n--- Statistics ---")
print(f"Average Final Grade: {df['Final_Grade'].mean():.2f}")
print(f"Median Final Grade: {df['Final_Grade'].median():.2f}")
print(f"Highest Final Grade: {df['Final_Grade'].max():.2f}")
print(f"Lowest Final Grade: {df['Final_Grade'].min():.2f}")
print(f"Standard Deviation: {df['Final_Grade'].std():.2f}")
```
